```python
import numpy as np 
import matplotlib.pyplot as plt 
from data_loaders import DataLoader
from statistical_methods import StatMethods
```

```python
def accumulate_data(month_data):
    return month_data["f"].to_numpy
```

```python

dataloader = DataLoader("D:/group project data/frequency_data", [x for x in range(2016, 2026)], [x for x in range(1, 13)])
def histogram(month_data):
    data = month_data["f"].to_numpy()
    bins = np.linspace(48, 52, 101)
    hist, _ = np.histogram(data, bins=bins)
    return hist


result = dataloader.run_on_data(histogram) #we need to write a new method 
#result is going to contain an array of histograms 
```

```python
total_histogram = np.zeros(100, dtype=np.int64)
for x in result:
    total_histogram += x 

```

```python
len(total_histogram[20:-20])
```

```python
import matplotlib.pyplot as plt 
bins = np.linspace(48, 52, 101)
plt.bar(
    bins[20:-21],
    total_histogram[20: -20],
    width=np.diff(bins[20:-20]),
    align="edge",
    edgecolor="black"
)
plt.xlabel("value")
plt.ylabel("frequency ")
plt.show() 
```
